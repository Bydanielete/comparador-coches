```python
numeros = pd.to_numeric(phev["rangeA"], errors="coerce")
phev.loc[numeros.isna(), ["make", "model", "year", "rangeA"]]
```

```python
phev["rangeA"] = pd.to_numeric(phev["rangeA"])
phev["cylinders"] = phev["cylinders"].astype(int)
phev.dtypes
```

```python
phev.columns.tolist()
```

```python
phev["consumo_l100km"] = (235.215 / phev["comb08"]).round(1)
phev["consumo_kwh100km"] = (phev["combE"] / 1.609344).round(1)
phev["co2_gkm"] = (phev["co2TailpipeGpm"] / 1.609344).round(0).astype(int)
phev["autonomia_km"] = (phev["rangeA"] * 1.609344).round(0).astype(int)
```

```python
phev[["comb08", "consumo_l100km", "rangeA", "autonomia_km"]].head()
```

```python
nombres = {
    "id": "id",
    "make": "marca",
    "model": "modelo",
    "year": "anio",
    "VClass": "carroceria",
    "consumo_l100km": "consumo_l100km",
    "consumo_kwh100km": "consumo_kwh100km",
    "autonomia_km": "autonomia_km",
    "charge240": "carga_horas",
    "co2_gkm": "co2_gkm",
    "displ": "cilindrada_l",
    "cylinders": "cilindros",
    "drive": "traccion",
    "trany": "transmision",
}

coches = phev[list(nombres.keys())].rename(columns=nombres)
coches.head()
```

```python
coches.to_json("../datos/coches.json", orient="records", force_ascii=False, indent=2)
```

## explicacion codigo
"../datos/coches.json": los dos puntos significan “sube una carpeta”. Como el notebook está en scripts/, sube a la raíz del proyecto y entra en datos/.
orient="records" genera una lista de objetos, uno por coche. Es el formato que mejor entiende JavaScript.
force_ascii=False guarda las tildes tal cual en lugar de convertirlas en códigos como \u00ed.
indent=2 lo deja con sangría, legible para un humano.
